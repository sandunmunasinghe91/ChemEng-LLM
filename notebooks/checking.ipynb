```python
import csv
import random
from pathlib import Path

REPORT_PATH = Path("../reports/arxiv_quality_report.csv")
PROCESSED_DIR = Path("../data/processed/arxiv_latex")

# Read report
with REPORT_PATH.open(
    "r",
    encoding="utf-8",
) as file:
    reader = csv.DictReader(file)
    rows = list(reader)

# Keep only papers flagged for broken references
broken_rows = [
    row
    for row in rows
    if "broken_references" in row["failed_checks"]
]

print(f"Found {len(broken_rows)} papers with broken references")

# Randomly choose up to 10 papers
sample = random.sample(
    broken_rows,
    k=min(10, len(broken_rows)),
)

for row in sample:

    file_path = PROCESSED_DIR / row["filename"]

    text = file_path.read_text(
        encoding="utf-8",
        errors="replace",
    )

    print("\n" + "=" * 80)
    print("FILE:", row["filename"])
    print("REASON:", row["reasons"])
    print("=" * 80)

    # Show first 1500 characters
    print(text[:1500])
```
